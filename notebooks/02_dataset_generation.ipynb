```python
import pandas as pd
import numpy as np
import random
```

```python
NUM_ROWS = 3000

routes = [f"R{i}" for i in range(1, 21)]   # 20 routes
hours = list(range(24))
day_types = ["Weekday", "Weekend"]
weather_conditions = ["Clear", "Rain"]
```

```python
def assign_crowd_level(hour, day_type, weather, is_holiday):
    score = 0

    if 8 <= hour <= 10 or 17 <= hour <= 20:
        score += 2

    if day_type == "Weekend":
        score -= 1

    if weather == "Rain":
        score += 1

    if is_holiday == 1:
        score -= 1

    # 🔥 ADD NOISE (real-world uncertainty)
    score += random.choice([-1, 0, 0, 1])

    if score <= 0:
        return "Low"
    elif score == 1:
        return "Medium"
    else:
        return "High"
```

```python
data = []

for _ in range(NUM_ROWS):
    route = random.choice(routes)
    hour = random.choice(hours)
    day_type = random.choice(day_types)
    weather = random.choice(weather_conditions)
    is_holiday = random.choice([0, 1])

    crowd = assign_crowd_level(hour, day_type, weather, is_holiday)

    data.append([
        route,
        hour,
        day_type,
        is_holiday,
        weather,
        crowd
    ])
```

```python
columns = [
    "route_id",
    "hour",
    "day_type",
    "is_holiday",
    "weather",
    "crowd_level"
]

df = pd.DataFrame(data, columns=columns)
df.head()
```
