# AI-Based Public Transport Crowding Prediction System

This notebook documents the problem definition, dataset strategy, feature selection,
and data logic for the Smart City crowd prediction project.



## Problem Definition

Urban public transport systems often face unpredictable crowding, causing inconvenience
and safety concerns for commuters. This project aims to predict public transport crowd
levels using machine learning.

The problem is formulated as a multi-class classification task where the model predicts
whether a route will experience Low, Medium, or High crowd levels based on temporal,
environmental, and route-based features.


## Project Objective

The objective of this project is to build an AI-driven system that can accurately predict
crowd levels in public transport systems. The solution is intended to help commuters plan
their travel better and assist transport authorities in managing congestion efficiently.


## Dataset Strategy

Due to limited availability of open and real-time public transport crowd datasets,
a synthetic dataset will be generated using realistic assumptions.

The dataset simulates commuter behavior based on factors such as peak hours,
weekday versus weekend patterns, weather conditions, and holidays. This approach
ensures the data reflects real-world transport dynamics while remaining scalable.


## Feature Selection

The following features are selected for predicting crowd levels:

- route_id: Unique identifier for the transport route
- hour: Hour of the day (0–23)
- day_type: Indicates whether the day is a weekday or weekend
- is_holiday: Binary indicator (0 = No, 1 = Yes)
- weather: Weather condition (Clear or Rain)
- crowd_level: Target variable indicating crowd intensity (Low, Medium, High)


## Crowd Level Logic

Crowd levels are generated based on realistic transport behavior patterns:

- Peak hours (8–10 AM and 5–8 PM) generally experience higher crowd levels
- Weekends usually have lower crowd except during evening hours
- Rainy weather increases public transport usage, leading to higher crowd levels
- Holidays alter regular commuting patterns and are treated separately


## System Architecture

        ↓

        ↓

        ↓

        ↓
